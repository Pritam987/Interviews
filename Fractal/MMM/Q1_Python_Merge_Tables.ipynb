```python
import pandas as pd
file_path = 'Python Questions_IME interview Round.xlsx'
xls = pd.ExcelFile(file_path, engine='openpyxl')

# Load the three tables from the 'PyQ1' sheet
df_sales = pd.read_excel(xls, sheet_name='PyQ1', usecols="A:B", nrows=13)
df_tv_grps = pd.read_excel(xls, sheet_name='PyQ1', usecols="F:G", nrows=6)
df_video_impressions = pd.read_excel(xls, sheet_name='PyQ1', usecols="K:L", nrows=7)

# Rename columns for merging
df_sales.columns = ['Timestamp', 'Sales']
df_tv_grps.columns = ['Timestamp', 'TV GRPs']
df_video_impressions.columns = ['Timestamp', 'Video Impressions']

# Merge the three tables
merged_df = pd.merge(df_sales, df_tv_grps, on='Timestamp', how='outer')
merged_df = pd.merge(merged_df, df_video_impressions, on='Timestamp', how='outer')

# Display the merged table
print(merged_df.iloc[1:,:])
```
