```python
# Q1: Create a new column Age_Bucketed
def bucket_age(age):
    if age < 20:
        return '<20 years'
    elif 20 <= age <= 30:
        return '20-30 years'
    elif 30 <= age <= 40:
        return '30-40 years'
    elif 40 <= age <= 50:
        return '40-50 years'
    else:
        return '>50 years'

df['Age_Bucketed'] = df['Age'].apply(bucket_age)
df.head(5)
```

```python

# Q2: Create a new column to generate function codes for each function using department_df
df = df.merge(department_df, left_on='Function', right_on='Department', how='left')
df.drop(columns=['Department'], inplace=True)
df.head(5)
```

```python
# Q3: Create a new column to find outliers in the df: "Outlier", "Not an Outlier"
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

def detect_outlier(salary):
    if salary < lower_bound or salary > upper_bound:
        return 'Outlier'
    else:
        return 'Not an Outlier'

df['Outlier'] = df['Salary'].apply(detect_outlier)
df.head(5)
```
